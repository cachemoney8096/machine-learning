```python
import subprocess
import sys
import os

libraries = ["ultralytics"]

for lib in libraries:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "--upgrade", lib])
```

```python
CLUSTER = True
if not CLUSTER:
    HOME = "/Users/rubenhayrapetyan/Downloads/Code/FRC/machine-learning/2025-Coral_Detection_Training"
else:
    HOME = "/home/rhayrapetyan/2025-Coral_Detection_Training"
    os.makedirs(f"{HOME}/coral_detection", exist_ok=True)
```

```python
from ultralytics import YOLO

model = YOLO('yolov8s.pt')  # load a pretrained model (recommended for training)
if CLUSTER:
    os.environ['CUDA_VISIBLE_DEVICES'] = '0,1'
    results = model.train(
        data=f"{HOME}/2025 REEFSCAPE.v2i.yolov12/data.yaml",
        epochs=100,
        imgsz=640,
        batch=16,
        name="coral-detection-model-v1",
        device=[0,1],
        workers=8,
        project=f"{HOME}/coral_detection/"
    )    
else:
    results = model.train(
        data=f"{HOME}/2025 REEFSCAPE.v2i.yolov12/data.yaml",
        epochs=100,
        imgsz=640,
        batch=16,
        name="coral-detection-model-v1",
        workers=2,
        project=f"{HOME}/coral_detection/"
    )

metrics = model.val(split="test")
```
