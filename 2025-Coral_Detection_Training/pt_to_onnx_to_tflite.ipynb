```python
from ultralytics import YOLO

# Load your trained model
model = YOLO("coral_detection/coral-detection-model-v16/weights/best.pt")

# Export to TFLite using Ultralytics' built-in export
model.export(
    format='tflite',         # Export format
    imgsz=640,               # Image size (should match your training size)
    int8=False,              # Set to True for INT8 quantization (smaller, faster, slightly less accurate)
    # half=True,             # Uncomment for FP16 quantization
)

print("Export complete! TFLite model saved as: coral_detection/coral-detection-model-v16/weights/best_saved_model/best_float32.tflite")
```
